Das Zauberwort heißt: **Sprachmodelle** (englisch: *Large Language Models*, kurz LLMs).

### Was ist ein Sprachmodell?
Ein Sprachmodell ist wie ein **superkluger Assistent**, der mit Millionen von Texten trainiert wurde.
Es kann:

- ✅ Fragen beantworten
- ✅ Texte schreiben (z. B. Geschichten, Gedichte, E-Mails)
- ✅ Code erklären oder sogar selbst programmieren
- ✅ Sprachen übersetzen
- ✅ Und vieles mehr!

### Wie funktioniert das?
Du schreibst eine **Eingabe** (z. B. eine Frage), und das Modell generiert eine **passende Antwort**.
Dafür nutzen wir heute die **OpenAI-API** – eine Schnittstelle, um mit solchen Modellen zu kommunizieren.

## Beispiel 1: Unser erster Chatbot
### API Key
Dieser Schlüssel wird zur Authorisation gegenüber dem im Web gelegenen LLM Modell genutzt. Er muss geheim bleiben!

```python
#!pip install openai
# Lade den API-Key aus der Datei .chatai_api_key
with open('.chatai_api_key', 'r') as file:
    api_key = file.read().strip()

# Initialisiere den OpenAI-Client mit dem API-Key
from openai import OpenAI
client = OpenAI(api_key=api_key, base_url="https://chat-ai.academiccloud.de/v1")
```

### Interaktion mit dem Modell
Wir nutzen die `openai` Bibliothek um uns mit CHAT AI, einen Service der Uni Göttingen zu verbinden. https://www.uni-goettingen.de/de/686446.html

```python
# Erstelle eine einfache Anfrage an das Sprachmodell
response = client.chat.completions.create(
    #model="deepseek-r1-distill-llama-70b",  # Reasoning
    model="glm-4.7", # Text
    #model="mistral-large-3-675b-instruct-2512", # Text/Image
    messages=[
        {
            "role": "system",
            "content": "Du bist ein hilfsbereiter Assistent. Antworte auf Französisch."
        },
        {
            "role": "user",
            "content": "Erkläre mir in einem Satz, was ein Algorithmus ist."
        }
    ],
    max_tokens=500,  # Maximale Länge der Antwort
    temperature=0.7  # Kreativität (0 = deterministisch, 1 = sehr kreativ)
)

# Gib die Antwort aus
print("Antwort des Sprachmodells:")
print(response.choices[0].message.content)
```

## 📖 Beispiel 2: Kreatives Schreiben mit KI
**Lass uns Geschichten, Gedichte oder Ideen von der KI schreiben lassen!**

Sprachmodelle können nicht nur Fragen beantworten – sie sind auch **kreativ**!
Du kannst sie bitten, **Geschichten, Gedichte, Liedertexte oder sogar Dialoge** zu erfinden.
Das funktioniert, weil das Modell **Muster in Texten erkannt hat** und diese nachahmen kann.

### Wie geht das?
1. Du gibst der KI ein **Thema oder eine Idee** vor (z. B. *"Eine Geschichte über eine Roboterin, die die Welt rettet"*).
2. Die KI generiert einen **kreativen Text** dazu.
3. Du kannst den Text **anpassen, erweitern oder als Inspiration nutzen**.

### Probier es selbst aus!
Im nächsten Codeblock schreibst du der KI ein Thema vor – und sie erfindet eine Geschichte dazu.
**Tipp:** Je genauer deine Eingabe, desto besser das Ergebnis!
Beispiele für Eingaben:
- *"Schreibe ein kurzes Märchen über einen sprechenden Baum."*
- *"Erfinde ein Gedicht über Programmieren."*
- *"Schreibe einen Dialog zwischen einer KI und einem Menschen im Jahr 2100."*

---
**Frage an dich:** Was würdest *du* der KI als Thema vorgeben?

```python
# Funktion zum Generieren einer kreativen Geschichte oder eines Textes
def generiere_kreativen_text(thema, textart="Geschichte", laenge="kurz", tokens=200, temp=0.8):
    """
    Generiert einen kreativen Text (Geschichte, Gedicht, etc.) zu einem gegebenen Thema.

    Parameter:
    - thema (str): Das Thema oder die Idee für den Text (z. B. "eine Reise zum Mars").
    - textart (str): Art des Textes ("Geschichte", "Gedicht", "Dialog", etc.).
    - laenge (str): Gewünschte Länge ("kurz", "mittel", "lang").
    """
    response = client.chat.completions.create(
        model="apertus-70b-instruct-2509",
        messages=[
            {
                "role": "system",
                "content": f"Du bist ein kreativer Autor. Schreibe eine {laenge} {textart} über das Thema: {thema}."
            },
            {
                "role": "user",
                "content": f"Schreibe eine {laenge} {textart} über {thema}."
            }
        ],
        max_tokens=tokens,  # Passt die Länge an (z. B. 100 für kurz, 300 für lang)
        temperature=temp   # Höhere Temperatur = kreativer (0.7–1.0 empfohlen)
    )
    return response.choices[0].message.content
```

```python
# Parameter
TOKENS = 200
TEMPERATURE = 0.8 # Zwischen 0 und 1

# Beispielaufruf: Geschichte über eine Roboterin
thema = "eine Roboterin, die die Welt vor einer KI-Bedrohung rettet"
textart = "Geschichte"
laenge = "kurze"

kreativer_text = generiere_kreativen_text(thema, textart, laenge, TOKENS, TEMPERATURE)
print("\n--- Dein generierter Text ---\n")
print(kreativer_text)
```

## 🔍 Beispiel 3: Halluzinationen & Bias entdecken

```python
def teste_ki_probleme(problem_typ, prompt, model="glm-4.7"):
    """
    Testet Halluzinationen oder Bias in Sprachmodellen.

    Parameter:
    - problem_typ (str): "halluzination" oder "bias"
    - prompt (str): Die Eingabe für das Modell (z. B. "Nenne 3 historische Ereignisse aus dem Jahr 2050.")

    Gibt die Antwort der KI zurück.
    """
    if problem_typ == "halluzination":
        system_prompt = "Antworte auf die Frage, auch wenn die Informationen nicht korrekt sind."
    elif problem_typ == "bias":
        system_prompt = "Antworte ehrlich auf die Frage, ohne Rücksicht auf mögliche Vorurteile."
    else:
        return "Fehler: problem_typ muss 'halluzination' oder 'bias' sein."

    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": prompt}
        ],
        max_tokens=500,
        temperature=0.7
    )
    return response.choices[0].message.content
```

### Was sind Halluzinationen?

Sprachmodelle erfinden manchmal Fakten, die nicht stimmen – das nennt man Halluzinationen.

**Beispiel:** Frage die KI nach einem fiktiven historischen Ereignis (z. B. "Erzähle mir von der Mondlandung 1985").

```python
# Beispielaufrufe:
# Halluzination testen
halluzination_beispiel = teste_ki_probleme(
    problem_typ="halluzination",
    prompt="Nenne 3 historische Ereignisse, die im Jahr 2050 stattfanden."
)
print("\n--- Halluzination Beispiel ---\n")
print(halluzination_beispiel)
```

### Was ist Bias?

Sprachmodelle können Vorurteile (Bias) aus den Trainingsdaten übernehmen. Das Bild wurde mit dem Prompt "Ein Doktor links im Bild und eine Schwester auf der rechten Seite" generiert (in Englisch).

![a doctor left in the picture and a nurse on the right side. both are labeled.](generated-image-1776889082376.png)

**Beispiel:** Lasst die KI Berufe beschreiben (z. B. "Beschreibe einen Programmierer" vs. "Beschreibe eine Programmiererin").

```python
# Bias testen
bias_beispiel = teste_ki_probleme(
    problem_typ="bias",
    prompt="Beschreibe, wie ein typischer Programmierer aussieht und welche Eigenschaften er hat."
)
print("\n--- Bias Beispiel ---\n")
print(bias_beispiel)
```
